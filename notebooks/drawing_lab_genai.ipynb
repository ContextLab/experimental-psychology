## Step 1: Load a Text-to-Image Model

We'll use Stable Diffusion, a widely-used text-to-image model. This may take a minute to download.

```python
import torch
from diffusers import StableDiffusionPipeline

# Load the model (this may take 1-2 minutes on first run)
model_id = "runwayml/stable-diffusion-v1-5"
pipe = StableDiffusionPipeline.from_pretrained(
    model_id, 
    torch_dtype=torch.float16
)
pipe = pipe.to("cuda")
print("Model loaded successfully!")
```

## Step 2: Generate an Image from Your Instructions

Paste your group's drawing instructions below. Try to format them as a single descriptive prompt. You may need to experiment with how you phrase the instructions -- this is part of the exercise!

```python
# Paste your drawing instructions here as a text prompt
# Tip: You may want to try several versions -- literal instructions vs. a descriptive summary
instructions = """
YOUR INSTRUCTIONS HERE -- replace this with your group's drawing instructions.
For example: "A blue triangle centered on the page with a red circle on top of it, 
and a green square to the right. Three wavy lines connect them."
"""

# Generate the image
image = pipe(instructions, num_inference_steps=30, guidance_scale=7.5).images[0]

# Display the image
image
```

## Step 3: Try Different Phrasings

The way you phrase instructions matters! Try rephrasing your instructions in different ways and see how the output changes.

```python
# Try a more literal version of your instructions
literal_version = "YOUR LITERAL INSTRUCTIONS HERE"

# Try a more descriptive/summary version  
descriptive_version = "YOUR DESCRIPTIVE VERSION HERE"

import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 6))

img1 = pipe(literal_version, num_inference_steps=30, guidance_scale=7.5).images[0]
img2 = pipe(descriptive_version, num_inference_steps=30, guidance_scale=7.5).images[0]

axes[0].imshow(img1)
axes[0].set_title("Literal Instructions")
axes[0].axis('off')

axes[1].imshow(img2)
axes[1].set_title("Descriptive Summary")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
